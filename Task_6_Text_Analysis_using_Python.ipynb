```python
from collections import Counter
import string

text = "Python is easy. Python is useful."

words = text.lower().split()

clean_words = [
    word.strip(string.punctuation)
    for word in words
]

word_counts = Counter(clean_words)

repeated_words = {
    word: count
    for word, count in word_counts.items()
    if count > 1
}

print("Repeated Words:", repeated_words)
```

```python

from collections import Counter
import string

text = """
Artificial intelligence is changing the world.
Artificial intelligence helps people solve problems.
Python is useful for artificial intelligence.
"""

clean_text = text.lower()

# Count words
words = clean_text.split()
total_words = len(words)

# Count characters
total_characters = len(text)
characters_without_spaces = len(
    text.replace(" ", "").replace("\n", "")
)

# Count vowels
vowels = "aeiou"
vowel_count = sum(
    1 for char in clean_text if char in vowels
)

# Clean words
clean_words = [
    word.strip(string.punctuation)
    for word in words
]
clean_words = [word for word in clean_words if word]

# Find repeated words
word_counts = Counter(clean_words)
repeated_words = {
    word: count
    for word, count in word_counts.items()
    if count > 1
}

# Display report
print("===== TEXT ANALYSIS REPORT =====")
print("Total Words:", total_words)
print("Total Characters:", total_characters)
print("Characters Without Spaces:", characters_without_spaces)
print("Total Vowels:", vowel_count)
print("Repeated Words:", repeated_words)
```
